## The Sum rule

```python
from sympy import *
from scipy.misc import derivative
```

```python
x = Symbol('x')
Fx = 7*x**3 + x**2 + 1

Derivative(Fx, x).doit()
```

```python
derivative(lambda x: 7*x**3 + x**2 + 1, 1.0)
```

## Calculating Partial derivatives

```python
x, y, z = symbols('x, y, z', real=True)
f = x**2 + y**3 + z**4 + 5*x*y*z
f
```

```python
for var in (x, y, z):
    print(f"f_{str(var)} = {f.diff(var)}")
```
